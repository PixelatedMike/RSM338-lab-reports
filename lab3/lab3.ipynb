<h3>Data Preparation</h3>
a) & b)
First we load the "monthly" sheet from the given xlsx file into a pandas DataFrame, and parse the dates in the date column.

```python
import pandas as pd

df = pd.read_excel('PredictorData2024.xlsx')

# Convert first column to datetime from yyyymm format
df['yyyymm'] = pd.to_datetime(df['yyyymm'], format='%Y%m')

# Set as index
df.set_index('yyyymm', inplace=True)

# Format as months only
df.index = df.index.to_period('M')
```

c), d), e)
We then construct predictors from the raw data, and lag inflation such that the returns from month $t$ are predicted by the inflation value at month $t-2$.

```python
import numpy as np

df['ExRet'] = df['CRSP_SPvw'] - df['Rfree']
df['d/p'] = np.log(df['D12']) - np.log(df['Index'])
df['d/y'] = np.log(df['D12']) - np.log(df['Index'].shift(1))
df['e/p'] = np.log(df['E12']) - np.log(df['Index'])
df['d/e'] = np.log(df['D12']) - np.log(df['E12'])
df['tms'] = df['lty'] - df['tbl']
df['dfr'] = df['corpr'] - df['ltr']
df['dfy'] = df['BAA'] - df['AAA']
df['infl'] = df['infl'].shift(1)

print(df)
```
